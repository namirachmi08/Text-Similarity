## Jaccard

```python
# JACCARD
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import jaccard_score
from scipy.spatial.distance import pdist

# Mengambil teks hasil preprocessing
documents = df["Text_Clean"].fillna("").astype(str)

# Mengubah teks menjadi binary vector
vectorizer = CountVectorizer(binary=True)
X = vectorizer.fit_transform(documents).toarray()

# Menghitung jarak Jaccard antar semua dokumen
distances = pdist(X, metric='jaccard')

print("Jaccard distance:")
print(distances[:10])
```

```python
# Similarity dokumen 1 dan 2
Sd1d2 = jaccard_score(X[0], X[1])

# Similarity dokumen 1 dan 3
Sd1d3 = jaccard_score(X[0], X[2])

# Similarity dokumen 2 dan 3
Sd2d3 = jaccard_score(X[1], X[2])

print("Similarity Doc 1 - Doc 2:", Sd1d2)
print("Similarity Doc 1 - Doc 3:", Sd1d3)
print("Similarity Doc 2 - Doc 3:", Sd2d3)
```

```python
similarity = 1 - distances

print("Jaccard similarity:")
print(similarity[:10])
```

```python
!pip install contractions

import numpy as np
import pandas as pd

from nltk.stem import PorterStemmer
from nltk.corpus import stopwords

import sklearn
import nltk

nltk.download('stopwords')

from IPython.display import display, HTML

data = pd.read_csv("/content/drive/MyDrive/combined_data.csv", header='infer')

print("Total rows : ", len(data))

display(HTML(data.head().to_html()))
```

```python
print(data.columns.tolist())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

stemmer = PorterStemmer()
stop_words = set(stopwords.words("english"))

corpus = []

for description in data["description"].fillna("").to_list():
    tokens = description.lower().split()

    # Stemming dan stop-word removal
    tokens = [
        stemmer.stem(token)
        for token in tokens
        if token not in stop_words
    ]

    corpus.append(" ".join(tokens))

print(corpus[0])
```

```python
# Membuat TF-IDF index berdasarkan seluruh corpus
vectorizer = TfidfVectorizer(
    sublinear_tf=True,
    max_features=5000)

vectorizer.fit_transform(corpus)
print("Total features:", vectorizer.get_feature_names_out().shape)
vectorizer.get_feature_names_out()[:1000]
```

```python
from sklearn.metrics.pairwise import cosine_similarity

documents_tfidf_features = vectorizer.transform(corpus)

TopK = 10

# Query pencarian film
query = "a young man who discovers a mysterious secret"

# Mengubah query menjadi TF-IDF vector
query_tfidf_features = vectorizer.transform([query])

print("TF-IDF features for query", query_tfidf_features, "\n")

# Menghitung cosine similarity
similarities = cosine_similarity(
    documents_tfidf_features,
    query_tfidf_features
)

# Mengambil 10 dokumen dengan similarity tertinggi
top_indices = similarities[:, 0].argsort()[::-1][:TopK]

titles = []
descriptions = []
similarity_scores = []

for doc_index in top_indices:
    titles.append(data.iloc[doc_index]["title"])
    descriptions.append(
        str(data.iloc[doc_index]["description"]).replace("\n", " ")
    )
    similarity_scores.append(similarities[doc_index, 0])

results = pd.DataFrame({
    "Title": titles,
    "Description": descriptions,
    "Similarities": similarity_scores
})

display(HTML(results.to_html()))
```

```python
# Jaccard
stemmer = PorterStemmer()
stop_words = set(stopwords.words("english"))

corpus = []

for description in data["description"].fillna("").to_list():
    tokens = description.lower().split()

    tokens = [
        stemmer.stem(token)
        for token in tokens
        if token not in stop_words
    ]

    corpus.append(" ".join(tokens))

print(corpus[0])
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(
    binary=True,
    max_features=5000)

documents_binarized_features = vectorizer.fit_transform(corpus).toarray()

print("Total features:", vectorizer.get_feature_names_out().shape)

vectorizer.get_feature_names_out()[:1000]
```

```python
from scipy.spatial.distance import cdist

TopK = 10

query = "a young man who discovers a mysterious secret"

query_binarized_features = vectorizer.transform([query]).toarray()

# Menghitung Jaccard Similarity
similarities = 1 - cdist(
    documents_binarized_features,
    query_binarized_features,
    metric="jaccard"
)

# Mengambil Top 10
top_indices = similarities[:, 0].argsort()[::-1][:TopK]

titles = []
descriptions = []
similarity_scores = []

for doc_index in top_indices:
    titles.append(data.iloc[doc_index]["title"])
    descriptions.append(
        str(data.iloc[doc_index]["description"]).replace("\n", " ")
    )
    similarity_scores.append(similarities[doc_index, 0])

results = pd.DataFrame({
    "Title": titles,
    "Description": descriptions,
    "Similarities": similarity_scores
})

display(HTML(results.to_html()))
```
